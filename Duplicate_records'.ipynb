- interview question
- print duplicate records present in the data
- remove duplicate entrysolve using pyspark 
- spark sql

```python
data = [(1,'abc@gmail.com'),(2,'def@gmail.com'),(1,'abc@gmail.com')]
column=['id','name']

df =spark.createDataFrame(data,column)
df.show()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window
```

```python
df.groupBy('name').agg(count('*')).display()
```

```python
df.groupBy('name').count().filter(col('count')>1).display()
```

```python
# print duplicate records present in the data
df.withColumn('rnk',row_number().over(Window.partitionBy('id').orderBy('id'))).filter(col('rnk')>1).select('name').display()
```

```python
# remove duplicate entry
df.withColumn('rnk',row_number().over(Window.partitionBy('id').orderBy('id'))).filter(col('rnk')==1).select('name').display()
```

```python
df.createOrReplaceTempView('sample')
```

```python
%sql
select count(*),name 
from sample 
group by name 
having count(*)>1
```

## remove duplicates

```python
%sql
select distinct id,name from sample
```
