# Explode Function

![image_1785083286017.png](./image_1785083286017.png "image_1785083286017.png")

```python
data=[
    ('manish',['python','java','c']),
    ('rahul',['java','c']),
    ('kishore',['c++','c'])
      ]
schema=['name','language']

df=spark.createDataFrame(data=data,schema=schema)
df.display()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df.withColumn('language',explode(col('language'))).display()
```

# in sql

```python
df.createOrReplaceTempView('emp')
```

```python
%sql
select * from emp;
```

```python
%sql
select name , explode(language) as new_col from emp;
```
