```python
import pyspark
from pyspark.sql.window import Window
from pyspark.sql.functions import *
from pyspark.sql.types import *

user_flags.show()
flag_review.show()

df=user_flags.join(flag_review,on=user_flags.flag_id==flag_review.flag_id,how='inner')

df=df.filter(col('reviewed_outcome')=='APPROVED')

df=df.withColumn('username',concat(col('user_firstname'),lit(' '),col('user_lastname')))

df=df.groupBy(col('username')).agg(countDistinct(col('vieo_id')).alias('unique_video_count'))

df=df.withColumn('rnk',dense_rank().over(Window.orderBy(col('unique_video_count').desc())))

df=df.filter(col('rnk')==1).select(col('username'))

df.show()

df.toPandas()
```
