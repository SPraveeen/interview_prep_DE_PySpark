```python
import pyspark
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window

df=worker.join(title,on=(worker.worker_id==title.worker_ref_id),how='inner')
df=df.select(col('worker_id'),col('salary'),col('worker_title'))
df=df.withColumn('rnk',dense_rank().over(Window.orderBy(col('salary').desc())))
df=df.filter(col('rnk')==1).select(col('worker_title').alias('best_paid_title'))
df.show()
df.toPandas()
```
