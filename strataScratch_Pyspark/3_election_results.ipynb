```python
import pyspark
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window

voting_results.show()

df=voting_results.groupBy(col('voter')).agg(count(col('candidiate')).alais('vote_count'))

df=df.withColumn('vote_value',round(1*1.0/col('vote_count'),1))

combined_df=df.join(voting_results,on=voting_results.voter=df.voter,how='inner').select(col('candidate'),col('vote_value'))

result_df=combined_df.groupBy(col('candidate')).agg(round(sum(col('vote_value')),3).alias('total_vote_value'))

result_df.sort(col('total_value_value'),ascending=False).limit(1).select('candidate')

result_df.show()
```
